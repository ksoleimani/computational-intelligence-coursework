# Clustering

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn import cluster
from sklearn import datasets
from sklearn.preprocessing import MinMaxScaler,StandardScaler
from sklearn.cluster import DBSCAN
```

```python

nSamples = 500
seed = 30
noisyCircles = datasets.make_circles(n_samples=nSamples, factor=.5, noise=.05, random_state=seed)
noisyMoons = datasets.make_moons(n_samples=nSamples, noise=.05, random_state=seed)
blobs = datasets.make_blobs(n_samples=nSamples, random_state=seed)
rng = np.random.RandomState(seed)
noStructure = rng.rand(nSamples, 2), np.zeros(nSamples, np.int8)
randomState = 170
x , y = datasets.make_blobs(n_samples=nSamples, random_state=randomState)
transformation = [[0.6,-0.6],[-0.4,0.8]]
Xaniso = np.dot(x, transformation)
aniso = (Xaniso, y)
varied = datasets.make_blobs(n_samples = nSamples, cluster_std = [1.1, 2.5, 0.7] , random_state=randomState)
```

```python
fig , ax = plt.subplots(5 , 1 , figsize = (5,15))
color = np.array(["Blue","Red","Green"])
for (i,datasets) in enumerate([blobs , noisyMoons , noStructure , aniso , varied]):
    ax[i].scatter(datasets[0][:,0], datasets[0][:,1], c=color[datasets[1]])
```

## Preprocessing

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler
dataset = blobs
X, y = dataset
X = MinMaxScaler().fit_transform(X)
plt.scatter(X[:,0], X[:,1])
```

## Clustering

```python
from sklearn.cluster import KMeans
wcss = []
model = []
for i in range (1,11):
    model.append(KMeans(n_clusters= i ).fit(X))
    wcss.append(model[i-1].inertia_)

plt.plot(range(1,11), wcss)
plt.title('WCSS by Clusters')
plt.xlabel('Number of Clusters')
plt.ylabel('WCSSS')
plt.show
```

```python
yp = model[2].fit_predict(X)
cmap = plt.cm.jet
fig , ax = plt.subplots()
ax.scatter(X[:,0], X[:,1], c = yp, cmap = cmap)
```

# DBSCAN

```python
model = DBSCAN(eps = 0.05, min_samples = 10)
yp = model.fit_predict(X)
fig, ax = plt.subplots()
ax.scatter(X[:,0], X[:,1], c = yp +1, cmap = cmap)
```

# Conclusion

K-Means performed well on datasets with clearly defined, spherical clusters and low noise, such as the blobs dataset. It effectively partitions data when the number of clusters is known and the variance among clusters is relatively uniform. However, its major limitations are that it requires the number of clusters to be specified in advance and it struggles with irregularly shaped clusters, varying densities, and noise points.

On the other hand, DBSCAN (Density-Based Spatial Clustering of Applications with Noise) does not require predefining the number of clusters. It excels at identifying clusters of arbitrary shape and can effectively detect outliers as noise. For example, DBSCAN performed significantly better on datasets like noisy moons and aniso, where clusters were non-spherical and had varying densities. However, DBSCAN can be sensitive to its parameters (eps and min_samples) and may not perform well on datasets with highly varying cluster densities or where the optimal parameters are hard to determine.
